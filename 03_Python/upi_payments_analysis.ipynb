```python
# Step 5: Sort dataset by Month_Start_Date
df = df.sort_values("Month_Start_Date").reset_index(drop=True)

print(df.head())
```

```python
print("Last 5 rows after sorting:\n")
print(df.tail())
```

Month-on-Month growth


```python
df["MoM_Volume_Growth_%"] = (df["UPI_Volume_Mn"].pct_change() * 100).round(2)
df["MoM_Value_Growth_%"] = (df["UPI_Value_Cr"].pct_change() * 100).round(2)
```

```python
print(df[[
    "Month",
    "UPI_Volume_Mn",
    "UPI_Value_Cr",
    "MoM_Volume_Growth_%",
    "MoM_Value_Growth_%"
]].head(10))
```

Year-on-Year growth

```python
df["YoY_Volume_Growth_%"] = (df["UPI_Volume_Mn"].pct_change(periods=12) * 100).round(2)
df["YoY_Value_Growth_%"] = (df["UPI_Value_Cr"].pct_change(periods=12) * 100).round(2)
```

```python
print(df[[
    "Month",
    "UPI_Volume_Mn",
    "UPI_Value_Cr",
    "MoM_Volume_Growth_%",
    "MoM_Value_Growth_%",
    "YoY_Volume_Growth_%",
    "YoY_Value_Growth_%"
]].head(15))
```

```python
df["Month_Name"] = df["Month_Start_Date"].dt.strftime("%B")
df["Calendar_Quarter"] = df["Month_Start_Date"].dt.quarter
```

yearly summary table


```python
yearly_summary = df.groupby("Financial_Year").agg({
    "UPI_Volume_Mn": "sum",
    "UPI_Value_Cr": "sum",
    "Banks_Live": "mean",
    "Avg_Transaction_Value": "mean",
    "MoM_Volume_Growth_%": "mean",
    "MoM_Value_Growth_%": "mean"
}).reset_index().round(2)
```

```python
yearly_summary.columns = [
    "Financial_Year",
    "Total_UPI_Volume_Mn",
    "Total_UPI_Value_Cr",
    "Avg_Banks_Live",
    "Avg_Transaction_Value",
    "Avg_MoM_Volume_Growth_%",
    "Avg_MoM_Value_Growth_%"
]
yearly_summary
```

quarterly summary table






```python
quarterly_summary = df.groupby(["Financial_Year", "Quarter"]).agg({
    "UPI_Volume_Mn": "sum",
    "UPI_Value_Cr": "sum",
    "Banks_Live": "mean",
    "Avg_Transaction_Value": "mean",
    "MoM_Volume_Growth_%": "mean",
    "MoM_Value_Growth_%": "mean"
}).reset_index().round(2)
```

```python
quarterly_summary.columns = [
    "Financial_Year",
    "Quarter",
    "Total_UPI_Volume_Mn",
    "Total_UPI_Value_Cr",
    "Avg_Banks_Live",
    "Avg_Transaction_Value",
    "Avg_MoM_Volume_Growth_%",
    "Avg_MoM_Value_Growth_%"
]
quarterly_summary.head()
```
